# 실습 3: 내 공정 카드 만들기

### 용어 풀이 - 반도체 8대 공정

| 순서 | 공정 | 한 줄로 |
|---|---|---|
| 1 | 웨이퍼 제조 | 실리콘으로 얇은 원판(웨이퍼)을 만들어 표면을 거울처럼 다듬는다 |
| 2 | 산화 | 웨이퍼 표면에 얇은 산화막을 입혀 보호막이자 절연막으로 삼는다 |
| 3 | 포토 (노광) | 빛으로 회로 무늬를 웨이퍼에 찍는다. 도장 찍기에 가깝다 |
| 4 | 식각 (에칭) | 찍은 무늬를 따라 필요 없는 부분을 깎아낸다 |
| 5 | 증착·이온주입 | 얇은 막을 입히고(증착), 전기 성질을 바꿀 입자를 박아 넣는다(이온주입) |
| 6 | 금속배선 | 만들어진 소자들을 금속 선으로 연결한다 |
| 7 | EDS (Electrical Die Sorting) | 자르기 전에 칩 하나하나에 전기를 흘려 합격·불합격을 판정한다 |
| 8 | 패키징 | 합격한 칩을 잘라내 포장하고 외부와 연결되게 만든다 |

2~6번은 한 번으로 끝나지 않고 층을 쌓는 만큼 여러 번 반복된다.

EDS는 재료 분야에서 쓰는 EDS(에너지 분산 분광법, Energy Dispersive Spectroscopy)와 철자만 같고 전혀 다른 말이다. 여기서는 전기를 흘려 칩을 골라내는 검사를 뜻한다.

## Step 2. AI에게 카드 재료 받기

## 식각(Etching) 공정 정리

**1. 이 공정이 하는 일**

웨이퍼 위 포토레지스트로 보호되지 않은 부분의 박막을 화학적·물리적으로 제거해 원하는 회로 패턴을 새기는 공정.

**2. 주요 측정값**

| 항목 | 단위 | 설명 |
|---|---|---|
| Etch Rate (식각 속도) | Å/min 또는 nm/min | 단위 시간당 제거되는 막 두께 |
| CD (Critical Dimension, 임계치수) | nm | 식각 후 패턴의 선폭 |
| Etch Uniformity (균일도) | % | 웨이퍼 내 위치별 식각량 편차 |
| RF Power | W | 플라즈마 생성을 위한 RF 전력 |
| Chamber Pressure | mTorr | 식각 챔버 내부 압력 |

**3. 정상 범위를 벗어나면 생기는 문제**

| 항목 | 문제 |
|---|---|
| Etch Rate 이상 | 과다식각(단선·소자 손상) 또는 부족식각(잔막으로 인한 단락) |
| CD 이탈 | 트랜지스터 특성 변화 → 소자 성능 불량 |
| Uniformity 저하 | 웨이퍼 내 위치별 성능 편차, 수율 저하 |
| RF Power / Pressure 이상 | 플라즈마 불안정 → 식각 프로파일 불량, 마이크로 로딩 효과 |

**4. 불확실하거나 장비·현장마다 달라지는 부분**

- 습식 식각과 건식(플라즈마) 식각은 측정 항목·단위 체계가 다름 (위 표는 건식 식각 기준)
- 장비 제조사(Applied Materials, Lam Research 등)마다 센서 위치와 파라미터 명칭이 다를 수 있음
- SECOM 데이터셋의 590개 피처는 공정 단계별 의미가 공식적으로 공개되어 있지 않아, 위 항목명은 일반적인 반도체 식각 공정 지식에 근거한 추정임

## Step 3. 받은 답을 내 눈으로 검토하기

## Etch Rate(식각률) 더 알아보기

| 질문 | 내용 | 확실성 |
|---|---|---|
| 단위 | Å/min(옹스트롬/분) 또는 nm/min이 흔히 쓰임. 초 단위(nm/sec)로도 표기 | 확실 |
| 구하는 방법 | (식각 전 두께 − 식각 후 두께) ÷ 식각 시간 | 확실 |
| 현장에서 중요한 이유 | 목표 두께까지 깎이는 데 걸리는 시간(식각 종료 시점)을 정하는 기준값. 이 값이 예상과 다르면 과식각/부족식각으로 바로 이어짐 | 확실 |
| 다른 용도 | 웨이퍼 간·로트 간 값 편차를 보면서 장비 이상을 조기에 발견하는 지표로도 씀 | 확실 |
| 장비·회사별 차이 | 습식/건식(플라즈마), 가스 조성, 챔버 구조에 따라 절대값이 크게 달라짐 | 확실 |
| 정확한 목표 스펙 수치 | 회사·라인마다 다르고 보통 영업비밀이라 공개된 표준값을 찾지 못함 | 확실하지 않음 |
| SECOM 데이터셋 내 실제 매핑 | 590개 피처 중 어떤 열이 식각률에 해당하는지는 공식적으로 공개되지 않음 | 확실하지 않음 |

3. 여기서 재는 값 :

   - \[챔버 압력\] — \[mTorr\]
   - \[RF 파워\] — \[W\]
   - \[식각 균일도\] — \[%\]
4. 이 값이 흔들리면 : \[Etch Rate가 예상과 어긋나면 과식각·부족식각이 생기고, 압력·RF파워가 흔들리면 균일도가 나빠져 웨이퍼 안에서 CD(선폭)가 들쭉날쭉해진다\]
5. 내 전공·경력과 닿는 지점 : \[식품영양학 전공. 조리·발효 과정에서 온도·시간·농도 같은 조건이 결과물의 품질과 일관성을 좌우하는 구조가 익숙하다. 식각에서 압력·유량·온도를 정해진 범위 안에서 관리하는 것도 같은 논리로 느껴진다\]
6. 확인이 필요한 부분 : \["식각 시간"이 재는 값인지 정해 넣는 값인지 / 실제 목표 스펙 수치는 회사·라인마다 달라 공개된 표준값을 찾지 못함\]
