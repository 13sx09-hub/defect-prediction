## Step 1. 문제 정의

**한 문장** : 반도체 공정의 센서 측정값으로, 최종 검사 전에 불량 가능성이 높은 제품을 먼저 골라낸다

| 물음 | 내 답 |
|---|---|
| 누가 쓰나 | 공정 검사 담당자 |
| 무엇을 결정하나 | 이 제품을 정밀 재검사로 보낼지 말지 |
| 언제 결정하나 | 자동 측정이 끝나고 값이 표에 올라온 직후 |
| 이 판단이 늦으면 | 불량 제품이 다음 공정까지 넘어가고, 완제품에서 발견되면 그 묶음 전체를 다시 봐야 한다 |

## 면접용 30초 설명

제가 만든 건, 반도체를 만드는 과정(공정)에서 나오는 센서 측정값만 보고, 검사가 끝나기 전에 불량이 될 가능성이 높은 제품을 미리 골라내는 것입니다. 이걸 쓰는 사람은 공정 검사 담당자이고, 이 결과로 그 제품을 정밀 재검사, 그러니까 한 번 더 자세히 들여다보는 검사로 보낼지 말지를 정합니다. 판단은 검사 라인에 들어오기 전, 센서 측정값이 막 나온 직후에 이루어집니다. 만약 이 판단이 늦어지면, 불량 제품이 다음 공정까지 그대로 넘어가고, 나중에 고객사에서 발견되면 그때는 비용이 훨씬 더 커집니다.
